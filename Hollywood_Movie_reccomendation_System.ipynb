```python
import pandas as pd

movies = pd.read_csv('/tmdb_5000_movies.csv')
credits = pd.read_csv('/tmdb_5000_credits.csv')

print("Movies shape:", movies.shape)
print("Credits shape:", credits.shape)

movies.head()
```

```python
credits.columns = ['id', 'title', 'cast', 'crew']

movies = movies.merge(credits, on='title')

print("Combined shape:", movies.shape)
movies.head()
```

```python
print(movies['title'].duplicated().sum())
```

```python
movies = movies.drop_duplicates(subset='title')

print("Cleaned shape:", movies.shape)
```

```python
movies = movies[['id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]

movies.head()
```

```python
print(movies.columns.tolist())
```

```python
movies = movies[['id_x', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]

movies = movies.rename(columns={'id_x': 'id'})

movies.head()
```

```python
import ast

def extract_names(text):
    try:
        items = ast.literal_eval(text)
        return [item['name'] for item in items]
    except:
        return []

movies['genres'] = movies['genres'].apply(extract_names)
movies['keywords'] = movies['keywords'].apply(extract_names)

movies[['title', 'genres', 'keywords']].head()
```

```python
def extract_top_cast(text, limit=3):
    try:
        items = ast.literal_eval(text)
        return [item['name'] for item in items[:limit]]
    except:
        return []

def extract_director(text):
    try:
        items = ast.literal_eval(text)
        for item in items:
            if item['job'] == 'Director':
                return item['name']
        return ''
    except:
        return ''

movies['cast'] = movies['cast'].apply(extract_top_cast)
movies['crew'] = movies['crew'].apply(extract_director)

movies[['title', 'cast', 'crew']].head()
```

```python
movies['overview'] = movies['overview'].fillna('')
movies['overview'] = movies['overview'].apply(lambda x: x.split())

movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew'].apply(lambda x: [x])

movies[['title', 'tags']].head()
```

```python
movies['tags'] = movies['tags'].apply(lambda x: " ".join(x))
movies['tags'] = movies['tags'].apply(lambda x: x.lower())

movies['tags'][0]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')

vectors = vectorizer.fit_transform(movies['tags']).toarray()

print("Shape:", vectors.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend(movie_title):
    movie_title = movie_title.lower()
    matches = movies[movies['title'].str.lower() == movie_title]

    if matches.empty:
        return "Movie not found in database. Check spelling!"

    idx = matches.index[0]

    distances = similarity[idx]

    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    recommendations = []
    for i in movie_list:
        recommendations.append(movies.iloc[i[0]]['title'])

    return recommendations

# Try it!
print(recommend("Avatar"))
```

```python
print(recommend("The Dark Knight Rises"))
print(recommend("Spectre"))
```

```python
!pip install gradio -q

import gradio as gr

def recommend_ui(movie_title):
    result = recommend(movie_title)
    if isinstance(result, str):
        return result
    return "\n".join(result)

interface = gr.Interface(
    fn=recommend_ui,
    inputs=gr.Textbox(placeholder="Enter a movie name, e.g. Avatar"),
    outputs="text",
    title="Movie Recommendation System",
    description="Enter a movie you like, and get 5 similar movie suggestions!"
)

interface.launch()
```
